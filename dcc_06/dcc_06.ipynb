## 【要点まとめ】 表形式のファイル操作と探索的データ分析

### **表形式のファイル操作**

#### **Pandasライブラリと表形式データ**
これまでの演習では、主にテキストデータを扱ってきました。しかし、ビジネスや研究で使われるデータの多くは、Excelのシートのように行と列で構成された**表形式データ**です。Pythonでこのようなデータを効率的に扱うための中心的なライブラリが **Pandas** です。

Pandasは、データを操作・分析するための高性能なデータ構造を提供します。
- **Series (シリーズ)**: 1次元のデータ構造で、1つの列に相当します。
- **DataFrame (データフレーム):**: 2次元のデータ構造で、行と列を持つ表全体に相当します。Excelのシートやデータベースのテーブルのようなものです。

まず最初のステップとして、``pd.read_csv()`` や ``pd.read_excel()`` といった関数を使い、CSVファイルやExcelファイルの内容をDataFrameとしてプログラムに読み込みます。

### **探索的データ分析**

#### **探索的データ分析とは**
データを読み込んだら、すぐに集計や分析を始めるわけではありません。その前に、データがどのようなものであるか、その全体像や特徴を把握する作業が不可欠です。このプロセスを **探索的データ分析 (Exploratory Data Analysis, EDA)** と呼びます。

#### **Pandasによる探索的データ分析**
EDAは、データに対する「最初の挨拶」のようなものです。この段階で、データが持つ特性や、欠損値・外れ値といった潜在的な問題点を明らかにします。Pandasには、EDAを効率的に行うための便利な関数（メソッド）が多数用意されています。

- ``.head()`` / ``.tail()``: DataFrameの**先頭5行** (``.head()``) または**末尾5行** (``.tail()``) を表示します。データが正しく読み込めているか、どのような列があるかを素早く確認するのに使います。
- ``.shape``: DataFrameの**形状**、つまり「(行数, 列数)」をタプルで返します。データセットの規模（レコード数と特徴量の数）を把握できます。
- ``.info()``: DataFrameの**基本的な情報**を一覧表示します。各列のデータ型（数値、文字列など）、欠損していない値の数、メモリ使用量などが分かり、データの全体像を技術的な側面から把握するのに役立ちます。
- ``.describe()``: **数値データ**の列に対して、基本的な**要約統計量**を自動で計算します。
  - ``count``: データ数
  - ``mean``: 平均値
  - ``std``: 標準偏差
  - ``min``: 最小値
  - ``25%``, ``50%`` (中央値), ``75%``: 四分位数
  - ``max``: 最大値<br>
  これにより、各列のデータの中心的な傾向、ばらつきの度合い、値の範囲などを一目で把握でき、外れ値の存在を推測する手がかりにもなります。
- **可視化による分析**: 数値だけでは捉えきれないデータのパターンを直感的に理解するために、グラフによる可視化もEDAの重要な手法です。``matplotlib``や``seaborn``といったライブラリがよく使われます。
  - **ヒストグラム(Histogram)**: 一つの数値データの**分布**を可視化するグラフです。データがどの値の範囲に集中しているか、山が一つなのか複数あるのか、左右対称かなどを確認できます。外れ値の発見にも役立ちます。
  - **散布図(Scatter Plot)**: 二つの数値データの**関係性**を可視化するグラフです。点が右肩上がりに分布していれば「正の相関」、右肩下がりなら「負の相関」があると推測できます。他の点からポツンと離れた点は、外れ値である可能性を示唆します。
  
このように、探索的データ分析（EDA）は、要約統計量の算出とグラフによる可視化を組み合わせることで、本格的なデータ分析やモデル構築に進む前に、データへの深い理解を得るための重要な第一歩です。
