## 【要点まとめ】形態素解析、単語の正規化・ベクトル化

### **形態素解析**

私たちが普段使っている日本語の文章を、コンピュータが直接理解することは困難です。文章は単語の連なりで構成されていますが、日本語では英語のように単語がスペースで区切られていません（わかち書き）。そこで、文章を意味を持つ最小の単位（形態素）に分割し、それぞれの品詞（名詞、動詞、助詞など）を判別する処理が必要になります。この処理を**形態素解析**と呼びます。

例えば、「美しい花が咲いた」という文を形態素解析すると、以下のようになります。

| 表層形 | 品詞 | 原形 |
| :----- | :--- | :--- |
| 美しい | 形容詞 | 美しい |
| 花 | 名詞 | 花 |
| が | 助詞 | が |
| 咲いた | 動詞 | 咲く |

このように分割することで、コンピュータは文章を単語の集合として扱えるようになり、特定の単語の出現頻度を数えたり、後述するベクトル化を行ったりする準備が整います。形態素解析は、自然言語処理における最も基本的で重要な前処理の一つです。

---

### **単語の正規化・ベクトル化**

形態素解析によって分割された単語を、コンピュータがさらに扱いやすいように加工するのが「正規化」と「ベクトル化」です。

**単語の正規化:** データ分析の精度を高めるためには、同じ意味を持つ単語は同じ単語として扱う必要があります。例えば、「PC」と「ｐｃ」、「教える」と「教えた」は人間にとっては同じ意味ですが、コンピュータは別の文字列として認識してしまいます。このような表記の揺れを統一する処理が **正規化** です。アルファベットの大文字・小文字の統一や、動詞や形容詞を基本形（原形）に戻す **レンマ化（Lemmatization）** などがあります。日本語の場合、形態素解析を行うと同時に原形も取得できることが多いです。

**単語のベクトル化:** コンピュータは単語（文字列）を直接計算できません。そのため、単語を数値の集まり、すなわちベクトルに変換する必要があります。この処理を**ベクトル化**と呼び、主に以下の手法があります。

- **Bag-of-Words (BoW)** 最もシンプルな手法で、「単語の袋」という名前の通り、文章中の単語の出現順序や文法構造を無視し、各単語が何回出現したかだけを数えます。以下がベクトル化の例：
  $$1: \text{猫がネズミを追いかける},  2: \text{犬が猫を追いかける}$$
  この2つの文から作られる単語の辞書は $(猫, が, ネズミ, を, 追いかける, 犬)$ となります。これに基づき各文をベクトル化すると以下のようになります。
  $$\vec{v_1}=(1, 1, 1, 1, 1, 0) , \quad \vec{v_2}=(1, 1, 0, 1, 1, 1)$$
    
- **TF-IDF** BoWの考え方を発展させた手法で、単語の「重要度」を数値化します。 **TF (Term Frequency)** ある文書中で特定の単語がどれだけ多く出現するか。 **IDF (Inverse Document Frequency)** その単語が、他の多くの文書でどれだけ「珍しい」か。TF-IDFは、 **「ある文書では頻繁に出現するが、他の文書ではあまり出現しない単語」** ほど重要度が高いと判断します。これにより、「です」「ます」のような一般的すぎる単語の重要度を下げ、その文書のトピックを特徴づける重要な単語を浮かび上がらせることができます。
- **Word2Vec** ニューラルネットワークを用いた、より高度なベクトル化手法です。単語の出現頻度だけでなく、 **「その単語がどのような文脈（周囲の単語）で使われるか」** を学習します。これにより、単語の意味的な近さをベクトルの距離として表現できます。例えば、「王様」と「女王」のベクトルは近く、「机」のベクトルとは遠くなります。さらに、「王様 - 男性 + 女性 = 女王」のような意味の計算も可能になり、単語の持つニュアンスをより豊かに捉えることができます。

単語をベクトル（数値）に変換することで、コンピュータはテキストデータを数学的に扱えるようになります。例えば、機械学習モデルへの入力、文章の類似度計算、文章分類など高度な分析が可能になります。
