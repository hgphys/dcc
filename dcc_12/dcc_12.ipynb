## 【要点まとめ】 データ加工：スケーリング（標準化と正規化）

### **データ加工（標準化と正規化）**
前回はデータを「数値」に変換する加工を行いました。しかし、すべてのデータが「数値」であっても、まだ問題が残っている場合があります。それが「尺度（スケール）の違い」です。第12回では、この尺度を揃えるスケーリングについて学びます。

#### **なぜスケーリングが必要か？**

例えば、「身長」（例: 170cm）と「体重」（例: 60kg）という2つの数値データがあったとします。これらの数値は、**単位も範囲も全く異なります**。

多くの機械学習アルゴリズム（特に距離を計算するものや、勾配降下法を用いるもの）は、数値の「大きさ」に大きく影響されます。

- 170 と 60 をそのままモデルに入力すると、モデルは「身長の方が、体重よりも100以上も大きな値であり、より重要な特徴だ」と誤って解釈してしまいます。

- 「身長」（1.7m）と「体重」（60kg）のように単位を変えるだけで、モデルの学習結果が変わってしまうのは問題です。

**スケーリング**とは、このような単位や範囲の違いによる影響をなくし、すべての特徴量がモデルに対して対等に扱われるように、数値の尺度を揃える前処理のことです。

#### **スケーリングが特に重要なモデル**

- **距離ベースのモデル**: k-NN（k近傍法）、SVM（サポートベクターマシン）、クラスタリングなど
- **勾配降下法を用いるモデル**: 線形回帰、ロジスティック回帰、ニューラルネットワークなど

#### **手法1: 標準化 (Standardization)**

**標準化**は、データセットのすべての値を、**平均が 0**、**標準偏差が 1** になるように変換する手法です。

- **計算式**: ``(元の値 - 平均値) / 標準偏差``

- **特徴**: 変換後の値は、元のデータが平均からどれだけ標準偏差分離れているか（Zスコア）を示します。

- **外れ値への影響**: 外れ値が存在する場合、その外れ値も変換されますが、他のデータから離れた「外れ値」として残ります。**外れ値の影響を受けにくい**（= スケールが潰されにくい）ため、最も一般的に使われる手法です。

```python
# 標準化 (StandardScaler) のコード例
from sklearn.preprocessing import StandardScaler
import pandas as pd

df_scale = pd.DataFrame({'身長(cm)': [160, 170, 180],
                       '体重(kg)': [50, 60, 70]})

# 1. StandardScalerのインスタンスを作成
scaler_std = StandardScaler()

# 2. .fit_transform() で学習と変換を同時に行う
df_standardized = scaler_std.fit_transform(df_scale)
df_standardized
```

#### **手法2: 正規化 (Normalization / Min-Max Scaling)**

**正規化**は、データセットのすべての値を、**0から1の範囲**（または-1から1）に収まるように変換する手法です。

- **計算式**: ``(元の値 - 最小値) / (最大値 - 最小値)``

- **特徴**: 最小値が0に、最大値が1に対応付けられます。

- **外れ値への影響**: 「最小値」と「最大値」を計算に使うため、**外れ値の影響を非常に強く受けます**。例えば、身長データに「500cm」という外れ値が1つあるだけで、他のすべてのデータが0に近い非常に小さな範囲に押し込められてしまいます。

```python
# 正規化 (MinMaxScaler) のコード例
from sklearn.preprocessing import MinMaxScaler

df_scale = pd.DataFrame({'身長(cm)': [160, 170, 180],
                       '体重(kg)': [50, 60, 70]})

# 1. MinMaxScalerのインスタンスを作成
scaler_minmax = MinMaxScaler()

# 2. .fit_transform() で学習と変換を同時に行う
df_normalized = scaler_minmax.fit_transform(df_scale)
df_normalized
```

### **まとめ**

**実践的応用**:
- **スケーリング**は、異なる単位を持つ特徴量（例: ``価格``と``販売個数``、``年齢``と``年収``）を同時に機械学習モデルに入力する際には、ほぼ必須の前処理となります。
- 特にニューラルネットワークやSVMなどのアルゴリズムでは、スケーリングの有無がモデルの収束速度や最終的な精度に決定的な影響を与えます。

**ポイント**:
- **目的**: スケーリングの目的は、特徴量間の「単位（尺度）」の違いをなくし、すべての特徴量を対等に扱うことです。
- **使い分け**:は、数値データのノイズを減らし、モデルが傾向を掴みやすくするために行います。
  - **標準化 (StandardScaler)**: データに外れ値が含まれている可能性がある場合や、アルゴリズムがデータの分布を正規分布（釣鐘型）と仮定している場合に適しています。**最も汎用性が高く、第一選択肢**となります。
  - **正規化 (MinMaxScaler)**: データの最小値と最大値が明確に決まっており、外れ値が存在しないことがわかっている場合に有効です。画像処理（ピクセル値を0〜1に収める）などでよく使われます。

### dcc_12

「年齢」と「年収」というスケールが極端に異なる2つの特徴量を使って、「商品の購入有無(``Purchased``)」を予測する分類モデルを構築します。数値データの特徴量の尺度（スケール）を揃える2つの主要な手法、「標準化」と「正規化」が、機械学習モデルの予測性能にどのような影響を与えるかを、スケーリングを行わない場合と比較・考察した、スケーリング手法考察レポートを完成させてください。

```python
# --- ライブラリのインポート ---
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import accuracy_score
import numpy as np

# --- データの読み込み ---
filepath = 'data_12/data_12.csv'
df = pd.read_csv(filepath)
print("--- 元のデータ (先頭5行) ---")
display(df.head())

# 特徴量X（予測に使うもの）と 目的変数y（予測したいもの）を分離
X = df[['Age', 'Income']]
y = df['Purchased']

# 訓練データとテストデータに分割
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print("--- 元のデータ (X_trainの先頭5行) ---")
display(X_train.head())
print(f"\n訓練データの形状: {X_train.shape}")
print(f"テストデータの形状: {X_test.shape}")
```

```python
# --- モデル1の学習と評価 ---
knn1 = KNeighborsClassifier(n_neighbors=5)
knn1.fit(X_train, y_train)
y_pred1 = knn1.predict(X_test)

acc_no_scaling = accuracy_score(y_test, y_pred1)
print(f"モデル1 (スケーリングなし) の正解率: {acc_no_scaling:.3f}")
```

```python
# --- 標準化の実行 --- 
scaler_std = StandardScaler()
# 訓練データで「学習(.fit)」と「変換(.transform)」を同時に行う
X_train_std = scaler_std.fit_transform(X_train)
# テストデータは、訓練データで学習した平均・標準偏差を使って「変換(.transform)」のみを行う
X_test_std = scaler_std.transform(X_test)

print("--- 標準化後のX_train (先頭5行) ---")
print(X_train_std[:5])

# --- モデル2の学習と評価 ---
knn2 = KNeighborsClassifier(n_neighbors=5)
knn2.fit(X_train_std, y_train)
y_pred2 = knn2.predict(X_test_std)

acc_standard = accuracy_score(y_test, y_pred2)
print(f"\nモデル2 (標準化) の正解率: {acc_standard:.4f}")
```

```python
# --- 正規化の実行 --- 
scaler_minmax = MinMaxScaler()
X_train_norm = scaler_minmax.fit_transform(X_train)
X_test_norm = scaler_minmax.transform(X_test)

print("--- 正規化後のX_train (先頭5行) ---")
print(X_train_norm[:5])

# --- モデル3の学習と評価 ---
knn3 = KNeighborsClassifier(n_neighbors=5)
knn3.fit(X_train_norm, y_train)
y_pred3 = knn3.predict(X_test_norm)

acc_minmax = accuracy_score(y_test, y_pred3)
print(f"\nモデル3 (正規化) の正解率: {acc_minmax:.4f}")
```

「年齢」と「年収」というスケールが極端に異なる2つの特徴量を使って、「商品の購入有無(``Purchased``)」を予測する分類モデルを、k-NN（k近傍法）で構築した。数値データの特徴量の尺度（スケール）を揃える2つの主要な手法、「標準化」と「正規化」が、機械学習モデルの予測性能にどのような影響を与えるかを、スケーリングを行わない場合と比較・考察した。
