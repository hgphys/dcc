## 【要点まとめ】 公開データ、APIの活用

### **公開データ**

#### **公開データとは**
**公開データ（オープンデータ）** とは、国や地方自治体、研究機関などが、誰でも自由に利用（加工、再配布など）できるように公開しているデータのことです。これらは機械判読に適した形式（CSVなど）で提供されることが多く、統計情報、地理情報、公共交通機関の時刻表など、多岐にわたります。
例:
- **e-Stat**: 日本の政府統計の総合窓口。国勢調査や経済センサスなど、様々な統計データが提供されています。
- **国土数値情報**: 国土交通省が提供する、日本の国土に関する地理空間データ。

これらのデータは信頼性が高く、そのまま分析に利用できることが多いですが、どこにどのようなデータがあるかを探し、ダウンロードして利用する必要があります。

### **APIの活用**

#### **APIとは**
**API (Application Programming Interface)** とは、ソフトウェアやプログラム同士が情報をやり取りするための、決められた手順やデータ形式のルールのことです。これをWebサービスで利用する場合、**Web API**と呼びます。

APIは、レストランのウェイターに例えることができます。

- **あなた（プログラム）** は、厨房（サーバー）の中身を知らなくても、**ウェイター（API）** にメニュー（APIドキュメント）を見て **注文（リクエスト）** するだけで、欲しい料理（データ）を受け取ることができます。

前回のスクレイピングが「レストランの壁を壊して無理やり料理を取ってくる」非公式な方法だとすれば、APIは「決められたルールに従って、サービス提供者と円滑にデータをやり取りする」公式な方法です。

#### **APIを利用するメリット**
- **安定性**: スクレイピングと違い、Webサイトのデザイン変更（HTML構造の変更）の影響を受けません。提供者が保証する限り、安定してデータを取得できます。
- **効率性**: 必要なデータが、プログラムで扱いやすい構造化された形式（主に**JSON**）で提供されるため、HTMLを解析する手間が省けます。
- **正当性**: サービス提供者が公式に許可した方法であるため、利用規約違反などの倫理的な問題を心配する必要がありません。（ただし、APIごとの利用規約やリクエスト回数制限は守る必要があります。）

#### **JSON形式とは**
APIを通じてデータをやり取りする際に、最も広く使われているデータ形式が**JSON (JavaScript Object Notation)** です。JSONは、Pythonの辞書型のように、**キー（項目名）と値（データ）** のペアでデータを表現します。

JSONの例：
```json
{
  "productName": "高機能マウス",
  "price": 5000,
  "inStock": true,
  "specs": {
    "dpi": 1600,
    "buttons": 5
  }
}
```

このように階層構造を持つデータを簡潔に表現できるため、プログラムでの処理が非常に容易です。Pythonでは、``json``ライブラリを使って簡単に辞書型に変換できます。
