## 【要点まとめ】データの種類、データの収集と前処理の必要性

### **データの種類**

データは、その構造によって大きく3つに分類されます。

- **構造化データ (Structured Data)**
  行と列からなる表形式で管理されているデータです。Excelのシートやリレーショナルデータベースのように、構造が明確に定義されています。
  例：売上管理表、顧客名簿、アンケートの回答データ

- **半構造化データ (Semi-structured Data)**
  構造化データのように厳密な形式はありませんが、HTMLのタグやJSONのキーのように、データ内に構造を示す情報（メタデータ）が含まれているデータです。
  例：Webページ(HTML)、APIで送受信されるデータ(JSON, XML)

- **非構造化データ (Unstructured Data)** 特定の構造を持たないデータで、世の中のデータの約8割を占めると言われています。分析のためには、自然言語処理や画像認識といった高度な技術で内容を解釈し、構造化された情報を取り出す必要があります。
例：メール本文、SNSの投稿、会議の議事録、画像、音声、動画

---

### **データの収集と前処理の必要性**

実世界から収集された「生データ」は、そのままでは分析に使えないことがほとんどです。なぜなら、以下のような多くの問題を含んでいるからです。

- **欠損値**: 入力漏れなどにより、値が存在しない。
- **外れ値**: 測定ミスや入力ミスにより、他の値から極端に離れた値が混ざっている。
- **表記の揺れ**: 「日本」と「Japan」、「株式会社A」と「(株)A」のように、同じものを指すのに表現が異なる。
- **ノイズ**: データに含まれる無関係な情報や誤り。

データ分析の世界には **「Garbage In, Garbage Out (ゴミを入れたら、ゴミしか出てこない)」** という有名な言葉があります。これは、質の低いデータを分析しても、得られる結果は信頼性の低い無価値なものになってしまう、という意味です。

そのため、AIモデルの構築やデータ分析を行う前には、必ず **「データの前処理（データクレンジング）」** という工程が必要になります。前処理とは、収集した生データを分析可能な状態に「掃除」し「整える」作業です。この地道な作業が、最終的な分析結果やモデルの精度を大きく左右する、最も重要な工程の一つと言えます。
