## 【要点まとめ】 データ操作：列・行・要素の抽出と操作

### **列・行・要素の抽出と操作（フィルタリング）**

分析の第一歩は、巨大なデータセットから必要な部分だけを正確に抜き出すことです。Pandasでは、この操作を直感的に行うための強力な機能が提供されています。

#### **列の抽出**
``df['列名']`` または ``df[['列名1', '列名2']]`` のように、``[]``（ブラケット）を使って特定の列（Series）や複数の列（DataFrame）を抽出します。

```python
import pandas as pd
# サンプルデータを作成
data = {'カテゴリ': ['果物', '野菜', '肉', '果物'],
        '価格': [100, 150, 500, 120]}
df = pd.DataFrame(data)
```

```python
df
```

```python
# '価格'の列を抽出
df['価格']
```

#### **行の抽出（条件によるフィルタリング）**
分析で最もよく使う手法です。``[]`` の中に条件式（ブール値のSeries）を入れることで、条件に一致する行だけを抽出します。
  - ``df[df['年齢'] > 20]`` （「年齢」列が20より大きい行を抽出）
  - ``df[df['カテゴリ'] == 'Electronics']`` （「カテゴリ」列が'Electronics'である行を抽出）
  - 複数の条件は & (AND) や | (OR) で組み合わせます。（例: ``df[(df['年齢'] > 20) & (df['カテゴリ'] == 'Electronics')]``）

```python
df['価格'] >= 150
```

```python
# 例: '価格'が150以上の行を抽出
df[df['価格'] >= 150]
```

```python
# 例: '価格'が150以上　でかつ'カテゴリ'が'肉'の行を抽出
mask = (df['価格'] >= 150) & (df['カテゴリ'] == '肉')
df[mask]
```

#### **要素の抽出**
``.loc`` と ``.iloc``（要素の抽出）:より厳密な抽出を行うために、以下の2つのアクセサを使います。

```python
df
```

```python
df_loc = pd.DataFrame(data, index=['a', 'b', 'c', 'd'])
df_loc
```

``.loc[行ラベル, 列ラベル]``: ラベル名（インデックス名や列名）に基づいて抽出します。

```python
df_loc.loc['c','価格']
```

``.iloc[行番号, 列番号]``: 位置（0から始まる行番号・列番号）に基づいて抽出します。

```python
df_loc.iloc[2,1]
```

```python
df_loc.iloc[0:2, 0] # 0行目から2行目
```

#### **まとめ**

- **実践的応用**: 本日の行フィルタリングは、あらゆるシステムのロジックの根幹です。例えば、ECサイトにおける「評価4以上のレビューのみ表示する」機能や、金融システムにおける「取引額が100万円を超える怪しい取引を抽出する」といった不正検知ロジックは、すべてこの条件抽出の考え方に基づいています。
- **ポイント**:
  - SQLとの関連: Pandasの条件抽出 ``df[df['列名'] > 100]`` は、データベース言語SQLの ``WHERE 列名 > 100`` という操作に相当します。これはツール（PandasやSQL）が異なっても共通する、データ抽出の普遍的な概念です。
  - ``.loc`` vs ``.iloc``: ``.loc``（ラベルベース）と ``.iloc``（位置ベース）の厳密な使い分けは、意図した通りのデータを確実に抽出するために重要です。「スライス（:）を使った場合に終点を含むか否か」といった動作の違いを正確に理解することが、エラーの防止につながります。
