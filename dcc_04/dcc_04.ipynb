## 【要点まとめ】 スクレイピング

### **Webの基本技術**

Webサイトから情報を収集するスクレイピングを理解するためには、まずWebがどのように機能しているかを知る必要があります。Webは、情報を要求する側（**クライアント**）と、情報を提供する側（**サーバー**）が互いに通信することで成り立っています。

- **HTTP (HyperText Transfer Protocol)** <br>
  クライアント（私たちのWebブラウザなど）とサーバーが対話するための「言葉」や「ルール」です。プログラムから特定のURLにアクセスするのは、サーバーに対して「このページの情報をください」という**HTTPリクエスト**を送信していることになります。サーバーはそれに応えて、ページの情報を**HTTPレスポンス**として返します。

- **HTML (HyperText Markup Language)** <br>
  Webページの構造を定義するための言語です。サーバーから返される情報の本体は、主にこのHTMLで記述されています。HTMLは、文章の各部分を以下のようなタグや属性で意味付けしており、これらを手がかりにすることで、プログラムはWebページの中から必要な情報だけを正確に見つけ出すことができます。
  - タグ (``<h1>``, ``<p>``, ``<a>``など): その部分が「見出し」「段落」「リンク」といった、どのような要素であるかを示す。
  - 属性 (``id``, ``class``など): 各要素に固有のIDやグループ名（クラス名）を付け、要素を特定しやすくする。

---

### **スクレイピング**

**スクレイピング**とは、プログラムを用いてWebサイトから情報を自動的に抽出し、収集する技術のことです。人力でコピー＆ペーストする作業を自動化するイメージです。

#### **技術（HTTP・HTML・正規表現）**
スクレイピングの基本的な流れは以下の通りです。

1. **HTTPリクエストの送信**: Pythonの``requests``などのライブラリを使い、目的のURLにアクセスしてWebページの全情報（HTML）を取得します。
2. **HTMLの解析**: 取得したHTMLはただの文字列なので、``BeautifulSoup``などのライブラリを使って、プログラムが扱いやすいようにタグの構造を解析します。
3. **データの抽出**: 解析したHTMLの構造（タグや属性）を頼りに、目的のデータ（例: 特定の見出しや表の中身）を抽出します。単純なテキストパターンであれば、``dcc_02`` で学んだ**正規表現**を使って抽出することも可能です。

具体例: 例えば、以下のような商品リストページのHTMLの一部があったとします。
```html
<div class="product-item">
  <h2 class="product-name">高機能マウス</h2>
  <span class="price">5,000円</span>
</div>
<div class="product-item">
  <h2 class="product-name">静音キーボード</h2>
  <span class="price">8,000円</span>
</div>
```
このHTMLから商品名と価格を抽出する場合、プログラムは以下のように考えます。

1. まず、``class="product-item"``という属性を持つ``<div>``タグをすべて見つける。
2. 次に見つけ出した各``<div>``タグの中から、``class="product-name"``を持つ``<h2>``タグを探し、その中のテキスト（「高機能マウス」など）を商品名として取得する。
3. 同様に、``class="price"``を持つ``<span>``タグを探し、その中のテキスト（「5,000円」など）を価格として取得する。

このように、HTMLの構造を道しるべにしてデータを抜き出すのがスクレイピングの基本です。


#### **倫理・法的配慮（規約・robots.txt）**
スクレイピングは便利な技術ですが、無秩序に行うと相手のサーバーに大きな負荷をかけてしまい、迷惑行為となる可能性があります。そのため、以下の倫理的なルールを守る必要があります。

- **利用規約の確認**: Webサイトによっては、利用規約でスクレイピングを明確に禁止している場合があります。必ず事前に確認し、規約に従う必要があります。
- **robots.txtの尊重**: 多くのサイトは、``https://example.com/robots.txt`` というファイルで、どのページへの自動アクセスを許可／禁止するかを意思表示しています。これは法的な強制力はありませんが、従うのがマナーです。
- **サーバーへの配慮**: 短時間に大量のリクエストを送らないように、アクセス間隔を空ける（例: ``time.sleep()``を使う）ことや、誰がアクセスしているかを明示するために **User-Agent** を設定するなどの配慮が不可欠です。

#### **実際の活用（APIとの比較や保存方法）**
- **APIとの比較**: **API (Application Programming Interface)** とは、ソフトウェアやプログラム同士が情報をやり取りするための、決められた手順やデータ形式のルールのことです。もしWebサイトがAPIを提供している場合、常にAPIを利用すべきです。APIは提供者が意図した方法で安定的にデータを取得できますが、スクレイピングはサイトのHTML構造が変わると動かなくなる、脆い方法です。スクレイピングは **「APIが提供されていない場合の最終手段」** と位置づけられます。

- **保存方法**: スクレイピングで収集したデータは、そのままでは使いにくいため、適切な形式で保存する必要があります。表形式のデータであれば**CSVファイル**、より複雑な階層構造を持つデータであれば**JSONファイル**として保存するのが一般的です。
