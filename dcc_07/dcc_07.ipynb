## 【要点まとめ】 欠損値、外れ値の取り扱い

前回の探索的データ分析（EDA）でデータの全体像を掴んだら、次はその品質をさらに向上させるための具体的なデータクリーニングに入ります。今回は、データ分析の精度に大きな影響を与える「欠損値」と「外れ値」の発見方法と、その取り扱いについて学びます。

### **欠損値 (Missing Values)**

#### **欠損値とは**
**欠損値**とは、何らかの理由で値が入力されていない、データが空の状態のことです。Pandasでは、欠損値は ``NaN`` (Not a Number) という特別な値で表現されます。入力漏れ、意図的な無回答、システムの不具合など、様々な原因で発生します。欠損値が存在すると、平均値や合計などの計算が正しく行えなかったり、多くの機械学習アルゴリズムがエラーを起こしたりします。これらを放置すると、分析結果が歪められたり、そもそも分析が実行不可能になったりします。

#### **Pandasによる欠損値の特定**
Pandasは、データを操作・分析するための高性能なデータ構造を提供します。
- ``.isnull()``: DataFrameの各要素が欠損値かどうかを、True/Falseのブール値で返します。
- ``.isnull().sum()``: 各列に欠損値がいくつあるかを合計して一覧表示します。データセット全体の欠損状況を素早く把握するための、最も基本的なコマンドです。

#### **欠損値の取り扱い**
主な対処法は「削除」と「補完」の2つです。
- **削除 (Deletion)** 欠損値を含む行または列をデータセットから完全に削除する方法です。
  - ``.dropna()``: このメソッドを使います。
  - 長所: 非常にシンプルで簡単。
  - 短所: 貴重なデータを失うことになります。欠損値の割合が多い場合にこの方法を使うと、分析に必要なデータが大幅に減ってしまう可能性があります。
- **補完 (Imputation)** 欠損値を何らかの別の値で埋める方法です。データを失わずに済みますが、どのような値で埋めるかが重要になります。
  - ``.fillna()``: このメソッドを使います。
  - 主な補完方法:
    - 代表値で補完: 列の平均値 (``.mean()``)、中央値 (``.median()``)、最頻値 (``.mode()``) などで埋めます。外れ値の影響を受けにくい中央値がよく使われます。
    - 固定値で補完: ``0`` や ``"不明"`` のような特定の固定値で埋めます。
    - 前後の値で補完: 時系列データなどで、直前の値 (``method='ffill'``) や直後の値 (``method='bfill'``) で埋める方法もあります。
  - **【重要】補完の注意点とデータの捏造**:欠損値の補完は、あくまで統計的な **「推定」** であり、元の「真実のデータ」ではありません。どのような方法で値を補完したかを記録・報告せずに、あたかも元からそのデータがあったかのように見せかけることは、データの捏造と見なされかねない、非常に危険な行為です。データクリーニングのプロセスは、常に透明性を保つことが重要です。

### **外れ値 (Outliers)**

#### **外れ値とは**
**外れ値**とは、他の大多数の値から極端に離れた値のことです。例えば、アンケートの年齢データに「200歳」と入力されていたり、商品の価格が「-500円」となっていたりするケースが該当します。入力ミスや測定エラーが主な原因ですが、時には不正行為などの「異常だが意味のあるデータ」である場合もあります。外れ値は、平均値や標準偏差といった統計量を大きく歪ませます。また、グラフを作成した際に、外れ値だけがポツンと離れて表示され、他のデータの分布が潰れて見えなくなってしまうなど、可視化にも影響を与えます。

#### **外れ値の特定方法**
- 可視化による発見
  - **箱ひげ図 (Box Plot)**: 外れ値を発見するための最も代表的なグラフです。「ひげ」の範囲から外れた点が、外れ値の候補としてプロットされます。
  - **散布図 (Scatter Plot)**: 他の点の集団から大きく孤立している点が、外れ値の候補となります。
- **統計的手法による発見（四分位範囲を使う）**<br>
  箱ひげ図の背景にある統計的なルールで、**四分位範囲 (IQR)** を用いて外れ値を定義する方法が一般的です。
  - **IQR** = 第3四分位数 (Q3) - 第1四分位数 (Q1)
  - **外れ値の範囲**: ``Q1 - 1.5 * IQR`` よりも小さい値、または ``Q3 + 1.5 * IQR`` よりも大きい値。

#### **外れ値の取り扱い**
- **削除**: 欠損値と同様に、外れ値を含む行を削除します。原因が明らかな入力ミスである場合に有効です。
- **修正**: もし正しい値が推測できる場合（例: 身長500cm → 170cmの入力ミス）、値を修正します。
- **変換**: 対数変換などを用いて、外れ値の影響を小さくする方法もあります。
- **維持**: 外れ値が、システムの不正利用や希少なイベントなど、分析上重要な意味を持つ「異常」である場合は、安易に削除せず、その原因を深掘りすることが重要です。


このように、欠損値と外れ値を適切に特定し対処することは、分析結果の信頼性を保証するための、データクリーニングにおける不可欠な工程です。

### dcc_07_02
``dcc_07_01`` に続いて、``UnitPrice``（単価）と``Quantity``（販売個数）それぞれのIQR法に基づいた外れ値の上限を計算し、散布図上に境界線として描画することで、異常な取引を視覚的に特定する。描画した散布図と算出された数値を観察し、外れ値の分析レポートを完成させてください。

```python
# --- 散布図とIQR境界線による外れ値の可視化 ---

# UnitPriceの外れ値境界を計算
Q1_price = df_filled_one['UnitPrice'].quantile(0.25)
Q3_price = df_filled_one['UnitPrice'].quantile(0.75)
IQR_price = Q3_price - Q1_price
upper_bound_price = Q3_price + 1.5 * IQR_price

# Quantityの外れ値境界を計算
Q1_qty = df_filled_one['Quantity'].quantile(0.25)
Q3_qty = df_filled_one['Quantity'].quantile(0.75)
IQR_qty = Q3_qty - Q1_qty
upper_bound_qty = Q3_qty + 1.5 * IQR_qty

print(f"UnitPriceの外れ値上限: {upper_bound_price:,.0f} 円")
print(f"Quantityの外れ値上限: {upper_bound_qty:,.1f} 個")
```

```python
# --- ライブラリのインポート ---
# matplotlib のインポート
import matplotlib.pyplot as plt

# 散布図の描画
plt.figure(figsize=(6, 6))
plt.scatter(x=df_filled_one['UnitPrice'], y=df_filled_one['Quantity'])

# 外れ値の境界線を引く
plt.axvline(x=upper_bound_price, color='r', linestyle='--', label=f'upper_bound_price')
plt.axhline(y=upper_bound_qty, color='g', linestyle='--', label=f'upper_bound_qty')

plt.xlabel('UnitPrice')
plt.ylabel('Quantity')
plt.grid(True)
plt.legend(loc='lower right')
plt.show()
```

#### **外れ値分析レポート**

- ``UnitPrice``と``Quantity``の散布図を観察したところ、データはグラフの ____ の領域に集中していることがわかる。
- IQR法を用いて計算した結果、``UnitPrice`` の外れ値上限は ____ 円、``Quantity``の外れ値上限は ____ 個と算出された。
- 散布図上でこの2つの境界線より右上にプロットされている ____ 件の点は、単価と数量の両方で異常に高い値を持つ外れ値であると判断できる。
