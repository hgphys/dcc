## 【要点まとめ】 データ操作：関数を用いた操作

### **関数を用いた操作（データ変換）**

既存のデータから新しいデータを作り出したり、値を一括で変換したりする際に、関数を適用します。

``.apply(関数)``:DataFrameの行ごと（``axis=1``）、または列ごと（``axis=0``）に関数を適用します。例えば、各行の「単価」と「数量」から「合計金額」という新しい列を作成する際に使えます。

````python
import pandas as pd
# サンプルデータを作成
df_apply = pd.DataFrame({'単価': [100, 200], '数量': [3, 5]})
df_apply
````

````python
df_apply["合計"] = df_apply["単価"] * df_apply["数量"]
df_apply
````

````python
# '単価'と'数量'を掛けて'合計'列を作る関数を定義
def calculate_total(row):
    return row['単価'] * row['数量']
````

````python
# .applyで行ごとに関数を適用
df_apply['合計'] = df_apply.apply(calculate_total, axis=1)
df_apply
````

``.map(辞書または関数)``: Seriesの各要素に対して、辞書や関数に基づいた変換を行います。「性別」列の 1 を 男性 に、2 を 女性 に一括変換する、といった処理で活躍します。

````python
# サンプルデータ（DataFrame）を作成
df_map = pd.DataFrame({'性別コード': [1, 2, 1, 1, 2]})
mapping_dict = {1: '男性', 2: '女性'}
df_map
````

````python
# サンプルデータ（DataFrame）を作成
df_map = pd.DataFrame({'性別コード': [1, 2, 1, 1, 2]})
mapping_dict = {1: '男性', 2: '女性'}

# DataFrameの '性別コード' 列（=Series）に対して .map を適用し、新しい '性別' 列を作成
df_map['性別'] = df_map['性別コード'].map(mapping_dict)
df_map
````

``lambda式（無名関数）``: ``apply`` や ``map`` と組み合わせて使う、その場限りの簡単な関数です。上記の``.apply()``の例は、``lambda``式を使うと以下のように1行で簡潔に記述できます。

````python
# .applyとlambda式を組み合わせて'合計'列を作成
df_apply = pd.DataFrame({'単価': [100, 200], '数量': [3, 5]})
df_apply['合計'] = df_apply.apply(lambda row: row['単価'] * row['数量'], axis=1)
df_apply
````

### **まとめ**

- **実践的応用**:データ変換は、AI・機械学習における **特徴量エンジニアリング（Feature Engineering）** の中核技術です。``.map()``は「男性/女性」といったカテゴリ変数を「1/0」のような数値に変換（エンコーディング）する際に必須です。また、``.apply()``は、既存の「売上」と「面積」の列から「単位面積あたりの売上」といった、モデルの予測精度を高めるための新しい特徴量（説明変数）を生成する際によく使われます。
- **ポイント**:
  - ``.apply()``の``axis``引数: ``axis=1``（行ごとに関数を適用）と``axis=0``（列ごとに関数を適用）の使い分けは、意図したデータ操作を行う上で重要な知識です。
  - ``lambda``式の理解: ``lambda``式を使いこなすことで、Pythonのコードをより簡潔かつ効率的に記述できます。``.apply(lambda x: ...)`` のような記述を見た際に、x に何が渡されているか（列全体なのか、行データなのか、単一の値なのか）を即座に理解できるかが重要です。
