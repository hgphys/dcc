## 【要点まとめ】自然言語処理とは、正規表現

### **自然言語処理とは**

自然言語処理（Natural Language Processing, NLP）とは、私たちが日常的に使っている言葉（自然言語）をコンピュータに処理・理解させるための一連の技術のことです。コンピュータは数値の計算は得意ですが、言葉の曖昧な意味や文脈をそのままでは理解できません。そこで、テキストデータを単語に分割したり、文の構造を解析したりすることで、コンピュータが扱える形式に変換する必要があります。

ECサイトのレビュー分析、迷惑メールフィルタ、チャットボット、機械翻訳など、身の回りの多くのサービスでこの技術が活用されています。データ分析の前処理においては、大量のテキストデータから有益な情報を抽出したり、レビューのポジティブ・ネガティブを判定したりするための基礎技術として非常に重要です。

---

### **正規表現**

**正規表現（Regular Expression）** とは、文字列の中から特定のパターンを持つ部分を見つけ出すための「ルールの記述法」です。例えば、「郵便番号（数字3桁-数字4桁）」や「メールアドレス」といった決まった形式の文字列を、複雑なルールを組み合わせることで的確に検索・抽出できます。

データクリーニングの文脈では、以下のような場面で絶大な効果を発揮します。

- 情報抽出: ユーザーの投稿から電話番号やURLだけを抜き出す。
- データ整形: 全角英数字を半角に統一する、不要な記号を削除する。
- マスキング: 個人情報保護のため、氏名や住所などを***のような伏字に置換する。

```\d（任意の数字）```や ```.（任意の一文字）```、```*（0回以上の繰り返し）```といった特殊な文字（メタ文字）を組み合わせてパターンを表現します。Pythonでは標準ライブラリ ```re``` を用いて、正規表現を利用した文字列操作を手軽に行うことができます。
